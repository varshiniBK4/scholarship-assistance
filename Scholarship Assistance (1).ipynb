```python
import json
import random

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import json

with open("intents.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print("Data reloaded successfully!")
print("Number of intents:", len(data["intents"]))
```

```python
patterns = []
tags = []
responses = {}

for intent in data["intents"]:
    tag = intent["tag"]

    responses[tag] = intent["responses"]

    for pattern in intent["patterns"]:
        patterns.append(pattern)
        tags.append(tag)

print("Training data prepared successfully!")
print("Total sample queries:", len(patterns))
print("Total intents:", len(responses))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2)
)

X = vectorizer.fit_transform(patterns)

print("TF-IDF vectorization completed successfully!")
print("Training data shape:", X.shape)
```

```python
def predict_intent(user_query):
    query_vector = vectorizer.transform([user_query])

    similarity_scores = cosine_similarity(query_vector, X)[0]

    best_index = similarity_scores.argmax()
    best_score = similarity_scores[best_index]

    if best_score < 0.25:
        return None, best_score

    return tags[best_index], best_score


print("Intent recognition function created successfully!")
```

```python
test_queries = [
    "Who can apply for scholarship?",
    "What documents do I need?",
    "How do I apply for the scholarship?",
    "When is the last date to apply?",
    "How much money will I receive?",
    "How can I check my application?",
    "Can I renew my scholarship?"
]

for query in test_queries:
    intent, score = predict_intent(query)

    print("Query:", query)
    print("Detected Intent:", intent)
    print("Similarity Score:", round(score, 2))
    print("-" * 55)
```

```python
query = "What documents are required for scholarship?"

intent, score = predict_intent(query)

if intent is not None:
    response = random.choice(responses[intent])

    print("User Query:", query)
    print("Detected Intent:", intent)
    print("Similarity Score:", round(score, 2))
    print("Chatbot Response:", response)

else:
    print("Sorry, I could not understand your question.")
```

```python
print("=" * 60)
print("       SCHOLARSHIP ASSISTANCE CHATBOT")
print("=" * 60)

print("You can ask about:")
print("Eligibility, Documents, Application, Deadline, Amount,")
print("Status, Renewal and Scholarship Assistance.")
print("Type 'bye' to exit.")
print()

while True:

    user_input = input("You: ").strip()

    if not user_input:
        print("Bot: Please enter a question.")
        print()
        continue

    intent, score = predict_intent(user_input)

    if intent is None:
        print("Bot: Sorry, I could not understand your question.")
        print("Bot: Please ask a scholarship-related question.")
        print()
        continue

    response = random.choice(responses[intent])

    print("Detected Intent:", intent)
    print("Similarity Score:", round(score, 2))
    print("Bot:", response)
    print()

    if intent == "goodbye":
        print("Chatbot ended.")
        break
```

```python
test_data = [
    ("Hello", "greeting"),
    ("Who can apply for scholarship?", "scholarship_eligibility"),
    ("What documents do I need?", "required_documents"),
    ("How can I apply?", "application_process"),
    ("When is the last date?", "application_deadline"),
    ("How much scholarship will I get?", "scholarship_amount"),
    ("How can I check my application?", "application_status"),
    ("Can I renew my scholarship?", "scholarship_renewal"),
    ("Who can help me?", "contact_help"),
    ("Bye", "goodbye")
]

print("CHATBOT TESTING")
print("=" * 70)

for query, expected in test_data:
    predicted, score = predict_intent(query)

    if predicted == expected:
        result = "PASS"
    else:
        result = "FAIL"

    print("Query:", query)
    print("Expected:", expected)
    print("Predicted:", predicted)
    print("Result:", result)
    print("-" * 70)
```

```python
correct = 0
total = len(test_data)

for query, expected in test_data:
    predicted, score = predict_intent(query)

    if predicted == expected:
        correct += 1

accuracy = (correct / total) * 100

print("Intent Recognition Accuracy:", round(accuracy, 2), "%")
```
